**Hypothesis 1.**

H0: There is no association between Days_Indoors and Growing_Stress

H1: There is association between Days_Indoors and Growing_Stress

````python
import pandas as pd
from scipy.stats import f_oneway # For performing one-way ANOVA
import seaborn as sns
import matplotlib.pyplot as plt

# Group the dataset by the 'Growing_Stress' column and extract 'Days_Indoors' values, dropping any NaN values
groups = [group['Days_Indoors'].dropna() for name, group in dataset.groupby('Growing_Stress')]

# Perform one-way ANOVA test
f_stat, p_value = f_oneway(*groups)

print(f"F-Statistic: {f_stat}")
print(f"P-Value: {p_value}")

# Evaluate the P-value to decide whether to reject the null hypothesis (H0)
# Null hypothesis (H0): There is no significant difference between the group means
if p_value < 0.05: # Threshold of 0.05 for statistical significance
    print("P-Value >= 0.05:H0 Rejected")

else:
    print("P-Value < 0.05: H0 Accepted")

# Create a boxplot to visualize the distribution of 'Days_Indoors' for each 'Growing_Stress' group
plt.figure(figsize=(8, 6))
sns.boxplot(x='Growing_Stress', y='Days_Indoors', data=dataset)
plt.title("Boxplot of Days_Indoors by Growing_Stress")
plt.show()
````

**Hypothesis 2.**

H0: There is no association between Days_Indoors and Mood_Swings

H1: There is association between Days_Indoors and Mood_Swings

````python
import pandas as pd
from scipy.stats import f_oneway
import seaborn as sns
import matplotlib.pyplot as plt

# Group the dataset by the 'Mood_Swings' column and extract 'Days_Indoors' values, dropping NaN values
groups = [group['Days_Indoors'].dropna() for name, group in dataset.groupby('Mood_Swings')]

f_stat, p_value = f_oneway(*groups)

print(f"F-Statistic: {f_stat}")
print(f"P-Value: {p_value}")

# Determine the outcome of the hypothesis test based on the P-value
# Null hypothesis (H0): There is no significant difference between the group mean
if p_value < 0.05:
    print("P-Value < 0.05:H0 Accepted")
else:
    print("P-Value >= 0.05: H0 Rejected")


plt.figure(figsize=(8, 6))
sns.boxplot(x='Mood_Swings', y='Days_Indoors', data=dataset)
plt.title("Boxplot of Days_Indoors by Mood_Swings")
plt.show()
````

# **Hypotheses 1**
H0: Gender has no association with Growing Stress.

 H1: Gender is associated with Growing Stress

````python
import pandas as pd
from scipy.stats import chi2_contingency

cat_var1 = 'Growing_Stress'
cat_var2 = 'Gender'

contingency_table = pd.crosstab(dataset[cat_var1], dataset[cat_var2])
print("Contingency Table:\n", contingency_table)

chi2, p, dof, expected = chi2_contingency(contingency_table)

print(f"Chi-Square Statistic: {chi2}")
print(f"P-Value: {p}")
print(f"Degrees of Freedom: {dof}")
print("Expected Frequencies:\n", expected)

if p < 0.05:
    print("P-Value >= 0.05: H0 Rejected")
else:
    print("P-Value < 0.05: H0 Accepted")
````

````python
from sklearn.metrics import accuracy_score
from sklearn.model_selection import RandomizedSearchCV
import numpy as np

# Function to run RandomizedSearchCV and check for overfitting
def randomized_search_with_overfitting_check(model, param_grid, X_train, y_train, X_test, y_test, n_iter=10):
    random_search = RandomizedSearchCV(estimator=model, param_distributions=param_grid, n_iter=n_iter, cv=10, n_jobs=-1, verbose=1, random_state=42)
    random_search.fit(X_train, y_train)
    best_model = random_search.best_estimator_
    best_params = random_search.best_params_

    # Training accuracy
    y_train_pred = best_model.predict(X_train)
    train_accuracy = accuracy_score(y_train, y_train_pred)

    # Test accuracy
    y_test_pred = best_model.predict(X_test)
    test_accuracy = accuracy_score(y_test, y_test_pred)

    # Check for overfitting by comparing training and test accuracies
    overfitting = "Yes" if train_accuracy - test_accuracy > 0.1 else "No"  # Overfitting if the difference is greater than 10%

    return best_model, best_params, train_accuracy, test_accuracy, overfitting

# Evaluate models with overfitting check using RandomizedSearchCV
for name, model in models.items():
    print(f"Training {name}...")

    best_model, best_params, train_accuracy, test_accuracy, overfitting = randomized_search_with_overfitting_check(
        model, param_grids[name], X_train, y_train, X_test, y_test, n_iter=10  # Adjust n_iter as needed
    )

    print(f"Best Parameters for {name}: {best_params}")
    print(f"{name} Train Accuracy: {train_accuracy}")
    print(f"{name} Test Accuracy: {test_accuracy}")
    print(f"Overfitting: {overfitting}\n")
````

Red Bars: Indicate a feature that increases the model's output (prediction).

Blue Bars: Indicate a feature that decreases the model's output (prediction).

Blue bars (long) = Strong negative contribution to the prediction

````python
from collections import Counter
from imblearn.over_sampling import SMOTE

# Apply SMOTE to balance the training data
smote = SMOTE(sampling_strategy='auto', random_state=42)
X_train_resampled, y_train_resampled = smote.fit_resample(X_train, y_train)

# Check the class distribution after applying SMOTE
print("Resampled class distribution:", Counter(y_train_resampled))


````

````python
import matplotlib.pyplot as plt
from sklearn.metrics import accuracy_score, classification_report, roc_auc_score, roc_curve
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from catboost import CatBoostClassifier
from lightgbm import LGBMClassifier
from xgboost import XGBClassifier
from sklearn.preprocessing import label_binarize
import pandas as pd

# Initialize models
models = {
    "Decision Tree": DecisionTreeClassifier(random_state=42),
    "Random Forest": RandomForestClassifier(random_state=42),
    "Gradient Boosting": GradientBoostingClassifier(random_state=42),
    "CatBoost": CatBoostClassifier(verbose=0, random_state=42),
    "LightGBM": LGBMClassifier(random_state=42),
    "XGBoost": XGBClassifier(use_label_encoder=False, eval_metric='logloss')
}

# Create a DataFrame to store the results
results = []

# Create the figure for ROC curves
plt.figure(figsize=(10, 8))

# For each model, fit, evaluate, and plot ROC Curve
for model_name, model in models.items():
    # Fit the model
    model.fit(X_train_resampled, y_train_resampled)

    # Predictions
    y_pred = model.predict(X_test)
    y_pred_prob = model.predict_proba(X_test)  # Probability for each class

    # Accuracy
    acc = accuracy_score(y_test, y_pred)

    # Classification Report (focusing on class 1 metrics)
    report = classification_report(y_test, y_pred, output_dict=True)
    avg_precision = report["1"]["precision"]  # Precision for class 1
    avg_recall = report["1"]["recall"]  # Recall for class 1
    avg_f1 = report["1"]["f1-score"]  # F1-score for class 1

    # ROC AUC for multi-class classification
    y_test_bin = label_binarize(y_test, classes=[0, 1, 2])  # Binarize for ROC curve
    roc_auc = roc_auc_score(y_test_bin, y_pred_prob, multi_class='ovr')

    # Append results to the list
    results.append([model_name, acc, avg_precision, avg_recall, avg_f1, roc_auc])

    # ROC Curve (Only for the overall model performance, no per-class ROC)
    fpr, tpr, _ = roc_curve(y_test_bin.ravel(), y_pred_prob.ravel())  # Flatten both to calculate ROC
    plt.plot(fpr, tpr, label=f'{model_name} (AUC = {roc_auc:.2f})')

# Convert results to a DataFrame and print
df_results = pd.DataFrame(results, columns=["Model", "Accuracy", "Precision", "Recall", "F1-score", "ROC AUC"])
print("\nSummary of Results:\n", df_results)

# Plotting the ROC Curve for all models
plt.plot([0, 1], [0, 1], 'k--', label='Random Classifier')  # Diagonal line for random classifier
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve for All Models')
plt.legend(loc='lower right')
plt.show()

````
